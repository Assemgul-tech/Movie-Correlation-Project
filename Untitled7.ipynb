```python
import requests
from bs4 import BeautifulSoup

# Получаем HTML-код главной страницы
url = "https://tengrinews.kz"
response = requests.get(url)
response.encoding = 'utf-8'  # Устанавливаем правильную кодировку
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем заголовки новостей
headlines = soup.find_all('span class'='main_news_top_item_meta')  # Класс может отличаться

# Выводим заголовки
print("🔹 Главные заголовки новостей:")
for i, headline in enumerate(headlines, start=0):
    print(f"{i}. {headline.get_text(strip=True)}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# Настройка опций для Chrome
options = Options()
options.add_argument('--headless')  # Запуск в фоновом режиме
options.add_argument('--disable-gpu')
options.add_argument('--no-sandbox')

# Инициализация драйвера
driver = webdriver.Chrome(options=options)

# Открытие страницы
driver.get("https://tengrinews.kz")
time.sleep(5)  # Ожидание загрузки страницы

# Получение HTML-кода страницы
html = driver.page_source
driver.quit()

# Парсинг с помощью BeautifulSoup
soup = BeautifulSoup(html, "html.parser")

# Поиск заголовков новостей
headlines = soup.find_all('a', class_='content_main_news_title')

# Вывод заголовков
print("Главные заголовки новостей:")
for i, headline in enumerate(headlines, 1):
    title = headline.get_text(strip=True)
    if title:
        print(f"{i}. {title}")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://tengrinews.kz/rss/all.rss"  # RSS-лента всех новостей
response = requests.get(url)
soup = BeautifulSoup(response.content, features="xml")

items = soup.find_all("item")

print("🔹 Главные заголовки новостей:")
for i, item in enumerate(items[:10], 1):  # первые 10 заголовков
    title = item.title.text
    print(f"{i}. {title}")
```
