```python
def get_full_name(first_name,last_name):
    full_name = f"{first_name} {last_name}"
    return full_name.title()
    
while True:
    print("Введите свое имя и фамилию:")
    f_name = input("Имя: ")
    if f_name == "quit":
        break
            
    l_name = input("Фамилия:")
    if f_name == "quit":
        break
  
    full_name = get_full_name(f_name, l_name)
    print(f"Привет {full_name}")
```

```python
def greet_users(names):
    for name in names:
        print(name.title())

usernames = ['kairat','samal','akniet']
greet_users(usernames)
```

```python
paints = ['phone','computer','horse']
completed = []

while paints:
    current = paints.pop()
    print(current)
    completed.append(current)

for i in completed:
    print(i)
```

```python
def print_paints(paints,completed):
    while paints:
        current = paints.pop()
        completed.append(current)

def print_completed(completed):
    for image in completed:
        print(image)

paints =  ['phone','computer','horse']
completed = []

print_paints(paints, completed)
print_completed(completed)
print(paints)

paints_2 = ['mouse','keyboard','flash']
completed_2 = []

print_paints(paints_2[:], completed_2)
print_completed(completed_2)
print(paints_2)
```

```python
def make_pizza(size, *adds):
    print(f"Нужна пицца с размером {size},в которой состав состоит из:")
    for add in adds:
        print(add)

make_pizza(21, 'cheeze')
print()
make_pizza(21, 'cheeze','tomato','potato')
```
