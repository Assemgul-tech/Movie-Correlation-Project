```python
def calculate_mean(*numbers):
    sum = 0
    for i in numbers:
        sum = sum + i
    return sum/len(numbers)
```

```python
 calculate_mean(5, 6, 7)
```

```python
 calculate_mean(10, 20, 30)
```
