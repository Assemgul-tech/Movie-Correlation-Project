```python
# Задача: создать список из квадратов четных чисел в диапазоне от 0 до 10
```

```python
number = int(input("введите число:"))

squares = []

for i in range(number + 1):
    if i % 2 == 0:
        squares.append(i ** 2)
print(squares)
```

```python
# Временная перемнная

counter = 0
```

```python
text =input("введите ваше предложение:")

split_text = text.split()

vowels = "aeiouyAEIOUY"

counter = 0

for i in range(len(split_text)):
    if split_text[i][0] in vowels:
        counter = counter + 1
        
print(counter)
```

```python
# Практическое задание: найти и вывести все простые числа в заданном пользователем диапазоне
```

```python
number = int(input("введите конечное значение:"))

dividers = []

for i in range(2, number):
    if number % i == 0:
        dividers.append(i)

print(len(dividers))

if len(dividers) > 0:
    print("Число составное:")
else:
    print("Число простое:")
```
