```python
x6 = np.random.randint(10, size = (3, 3))
x7 = np.random.randint(10, size = (3, 3))

np.concatenate([x6, x7])
```

```python
ver = np.concatenate((x6, x7), axis = 0)

print("Вертикаль")
print(ver)
```

```python
hor = np.concatenate((x6, x7), axis = 1)

print("Горизонталь")
print(hor)
```

```python
verv = np.vstack((x6, x7))

print("Vstack")
print(verv)  
```

```python
horh = np.hstack((x6, x7))

print("Hstack")
print(horh)
```

```python
mass = np.random.randint(10, size = (6,6))

print(mass)
split_mass, split_mass1 = np.vsplit(mass, [3])

print("Первый")
print(split_mass)

print("Второй")
print(split_mass1)
```

```python
y1, y2 = np.hsplit(mass, [3])

print("Первый")
print(y1)

print("Второй")
print(y2)
    
```
