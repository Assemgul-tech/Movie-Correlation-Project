```python
# 1. Функция Суммирования:
   Напишите функцию sum_numbers, которая принимает два числа как аргументы и возвращает их сумму.

```

```python
def sum_numbers(a, b):
    return a + b
print(sum_numbers(5, 7))
```

```python
# 2. Функция Проверки Четности:
   Создайте функцию is_even, которая принимает одно число и возвращает True, если число четное,
    и False, если нечетное.
```

```python
def is_even(number):
    return number % 2 == 0

print(is_even(10))  
print(is_even(7))  
```

```python
# 3. Функция Поиска Максимума:
   Реализуйте функцию find_max, которая принимает список чисел и возвращает самое большое число в списке.

```

```python
def find_max(numbers):
    return max(numbers) if numbers else None 

print(find_max([3, 7, 2, 9, 5]))  
```

```python
# 4. Функция Конвертации Температуры:
   Напишите функцию celsius_to_fahrenheit, которая конвертирует температуру из градусов Цельсия в градусы
    Фаренгейта.

```

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print(celsius_to_fahrenheit(0))  
print(celsius_to_fahrenheit(100))
```

```python
# 5. Функция Подсчета Гласных:
Создайте функцию count_vowels, которая принимает строку и возвращает количество гласных букв 
(а, е, и, о, у, ы, э, ю, я) в этой строке. Например, для строки "привет" функция должна вернуть 2.
```

```python
def count_vowels(text):
    vowels = "аеиоуыэюяАЕИОУЫЭЮЯ"
    return sum(1 for char in text if char in vowels)

print(count_vowels("привет"))  
print(count_vowels("здравствуйте"))  
```
