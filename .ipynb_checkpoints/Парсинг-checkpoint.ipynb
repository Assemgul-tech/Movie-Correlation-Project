```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = 'https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
table = soup.find('table',{'class:''forextable'})
data =[]

if table:
    for row in table.find_all('tr')[1:]:
        cols = row.find_all('td')
        currency = cols[0].text.strip()
        rate = cols[2].text.strip()
        data.append([currency, rate])
    df = pd.DataFrame(data, columns=['Currency', 'Rate'])
    print(df)
else:
    print("Таблица не найдена.Пожалуйста,проверьте имя класса и убедитесь, что структура сайта не изменилась")
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

# URL таблицы курсов валют
url = 'https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html'

# Запрос к сайту
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Поиск таблицы с курсами
table = soup.find('table', class_='forextable')  # класс может измениться

# Парсинг заголовков
headers = [th.text.strip() for th in table.find_all('th')]

# Парсинг строк
rows = []
for tr in table.find_all('tr')[1:]:
    cols = tr.find_all('td')
    if cols:
        rows.append([td.text.strip() for td in cols])

# Создание DataFrame
df = pd.DataFrame(rows, columns=headers)

# Вывод таблицы
print(df)
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = 'https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html'

response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Находим таблицу по классу
table = soup.find('table', class_='forextable')

# Заголовки задаём вручную
headers = [ 'Currency', 'Rate']

# Собираем строки
rows = []
for tr in table.find_all('tr')[1:]:
    cols = tr.find_all('td')
    if len(cols) == 2:
        row = [td.text.strip() for td in cols]
        rows.append(row)

# Создаём DataFrame
df = pd.DataFrame(rows, columns=headers)

# Пример: вывод первых строк
print(df.head())
```

```python
import requests
from bs4 import BeautifulSoup

# URL главной страницы
url = 'https://tengrinews.kz/'

# Отправляем GET-запрос
headers = {
    'User-Agent': 'Mozilla/5.0'
}
response = requests.get(url, headers=headers)

# Создаём объект BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем заголовки новостей — по классу заголовков
headlines = soup.find_all('span', class_='content-main-title')  # проверено по сайту

# Выводим заголовки
print("Главные заголовки новостей:")
for i, headline in enumerate(headlines, 1):
    print(f"{i}. {headline.text.strip()}")
```

```python
import requests
from bs4 import BeautifulSoup

# URL главной страницы
url = 'https://tengrinews.kz/'

# Отправляем GET-запрос
headers = {
    'User-Agent': 'Mozilla/5.0'
}
response = requests.get(url, headers=headers)

# Создаём объект BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем блок "Последние" (например, через его класс)
latest_block = soup.find('div', class_='feed-block feed-block_type_latest')

# Извлекаем заголовки из этого блока
if latest_block:
    headlines = latest_block.find_all('span', class_='content-main-title')
    print("Заголовки из раздела 'Последние':")
    for i, headline in enumerate(headlines, 1):
        print(f"{i}. {headline.text.strip()}")
else:
    print("Блок 'Последние' не найден.")
```

```python
import requests
from bs4 import BeautifulSoup

# Адрес главной страницы
url = 'https://tengrinews.kz/'

# Отправляем GET-запрос
headers = {
    'User-Agent': 'Mozilla/5.0'
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем блок с заголовками "Последние"
latest_news_block = soup.find('div', class_='main-feed__section--left')  # по структуре сайта

# Ищем все элементы новостей в этом блоке
if latest_news_block:
    news_items = latest_news_block.find_all('a', class_='link')  # ссылки на новости

    print("Заголовки новостей из блока 'Последние':")
    for i, item in enumerate(news_items, 1):
        title = item.get_text(strip=True)
        if title:
            print(f"{i}. {title}")
else:
    print("Блок 'Последние' не найден.")
```

```python
import requests
from bs4 import BeautifulSoup

# Главная страница
url = 'https://tengrinews.kz/'

# Заголовки HTTP-запроса
headers = {
    'User-Agent': 'Mozilla/5.0'
}

# Отправляем запрос
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем блок "Последние" (левая колонка)
latest_news_block = soup.find('div', class_='main-feed__section--left')

if latest_news_block:
    news_items = latest_news_block.find_all('a', class_='link')

    print("Заголовки новостей из блока 'Последние':\n")
    for i, item in enumerate(news_items, 1):
        title = item.get_text(strip=True)
        link = item['href']

        # Проверка на относительный путь
        full_url = link if link.startswith('http') else f'https://tengrinews.kz{link}'
        
        print(f"{i}. {title}\n   {full_url}")
else:
    print("Блок 'Последние' не найден.")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# Настройки браузера
options = Options()
options.add_argument("--headless")  # Без GUI
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

# Запуск драйвера
driver = webdriver.Chrome(options=options)
driver.get("https://tengrinews.kz/")

# Дать время на загрузку JS-контента
time.sleep(3)

# Получение HTML после загрузки
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')

# Закрыть браузер
driver.quit()

# Ищем блок с последними новостями
latest_news = soup.select('div.main-feed__section--left a.link')

# Вывод
print("Заголовки из блока 'Последние':\n")
for i, item in enumerate(latest_news, 1):
    title = item.get_text(strip=True)
    href = item['href']
    url = href if href.startswith('http') else f'https://tengrinews.kz{href}'
    print(f"{i}. {title}\n   {url}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(options=options)
driver.get("https://tengrinews.kz/")
time.sleep(5)  # Подольше ждём

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
driver.quit()

# Пробуем оба варианта селектора
latest_news = soup.select('div.main-feed__section--left a.link')
if not latest_news:
    latest_news = soup.select('div.latest-news a')

print(f"Найдено заголовков: {len(latest_news)}\n")

for i, item in enumerate(latest_news, 1):
    title = item.get_text(strip=True)
    href = item['href']
    url = href if href.startswith('http') else f'https://tengrinews.kz{href}'
    print(f"{i}. {title}\n   {url}")
```
