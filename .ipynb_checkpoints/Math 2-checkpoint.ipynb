```python
#1. Создайте функцию f(x) = x^6 + 10 * x^3 - 1000
```

```python
def f(x):
    return x**6 + 10 * x**3 - 1000
```

```python
#2. Создайте список x_points (или массив через numpy), в котором у нас лежат числа от -10 до 10
```

```python
import numpy as np

x_points = np.linspace(-10, 10, 1000)  # 1000 точек для плавного графика
```

```python
#3. Используйте x_points как аргумент для созданной функции f(x), и выведите на экран результаты функции для каждого значения 
# x_points, также сохраните результаты в y_points (cписок или массив)
```

```python
y_points = f(x_points)

# Выводим несколько значений
for x, y in zip(x_points[:10], y_points[:10]):  # первые 10 для примера
    print(f"f({x:.2f}) = {y:.2f}")
```

```python
#4. Покажите на графике x_points & y_points
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(x_points, y_points, label='f(x)')
plt.title('График функции f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#5. Найдите производную функции f(x) и создайте новую функцию на ее основе как f_diff(x)
```

```python
def f_diff(x):
    return 6 * x**5 + 30 * x**2
```

```python
#6 Чему равна производная при x = 10
```

```python
print(f"f'(10) = {f_diff(10)}")
```

```python
#7 Покажите на графике f(x) & f_diff(x)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)')
plt.plot(x_points, f_diff(x_points), label="f'(x)", linestyle='--')
plt.title('График функции f(x) и её производной f\'(x)')
plt.xlabel('x')
plt.ylabel('Значение')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#8 Найдите уравнение касательной линии в точке x = 9
```

```python
x0 = 9
y0 = f(x0)
slope = f_diff(x0)  # производная в точке 9

def tangent_line(x):
    return y0 + slope * (x - x0)
```

```python
#9 Покажите на графике функцию f(x) & касательную линию в точке x = 9
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)')
plt.plot(x_points, tangent_line(x_points), label='Касательная в x=9', linestyle='--')
plt.title('График функции и касательной')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```
