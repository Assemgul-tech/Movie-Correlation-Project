```python
class Restaurant:
    def __init__(self, restaurant_name, cuisine_type):
        self.restaurant_name = restaurant_name
        self.cuisine_type = cuisine_type
  
    def describe_restaurant(self):
        print(f"Название ресторана: {self.restaurant_name}, кухня: {self.cuisine_type}")
    
    def open_restaurant(self):
        print(f"Ресторан с названием: {self.restaurant_name} открыт")
    
    
restaurant = Restaurant("Luckee", "Asian")
    
print(restaurant.restaurant_name)
print(restaurant.cuisine_type)
    
restaurant.describe_restaurant()
restaurant.open_restaurant()
   
```

```python
class Restaurant():
    def __init__(self, name, cuisine):
        self.restaurant_name = name
        self.cuisine_type = cuisine
        self.number_served = 0
    
    
    def describe_restaurant(self):
        print(f"Название ресторана: {self.restaurant_name} кухня: {self.cuisine_type}")
    
    def open_restaurant(self):
        print(f"Ресторан: {self.restaurant_name} открыт")
        
    def set_number_served(self, number):
        self.number_served = number
   
        

restaurant_2 = Restaurant("Del Papa", "Italian")
print(restaurant_2.number_served)

restaurant_2.number_served = 2
print(restaurant_2.number_served)

restaurant_2.set_number_served(4)
print(restaurant_2.number_served)
```

```python
class Restaurant():
    def __init__(self, name, cuisine):
        self.restaurant_name = name
        self.cuisine_type = cuisine
        self.number_served = 0
    
    
    def describe_restaurant(self):
        print(f"Название ресторана: {self.restaurant_name} кухня: {self.cuisine_type}")
    
    def open_restaurant(self):
        print(f"Ресторан: {self.restaurant_name} открыт")
        
    def set_number_served(self, number):
        self.number_served = number
        
    def increment_number_served(self, increment):
        self.number_served += increment
    
   
        

restaurant_2 = Restaurant("Del Papa", "Italian")
print(restaurant_2.number_served)


restaurant_2.increment_number_served(10)
print(restaurant_2.number_served)
```

```python
class User:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        
    def describe_user(self):
        print(f"Имя: {self.first_name} Фамилия: {self.last_name}")
    
    
    def greet_user(self):
        print(f"Приветствую вас {self.first_name} {self.last_name}")
    
user1 = User("Kairat", "Omarov")
user2 = User("Valera", "Popov")
user3 = User("Saltanat", "Gazizqyzy")
    
user1.describe_user()
user1.greet_user()

user2.describe_user()
user2.greet_user()
    
user3.describe_user()
user3.greet_user()
```

```python
class User:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.login_attempts = 0 
    
    def describe_user(self):
        print(f"Имя: {self.first_name} Фамилия: {self.last_name}")
    
    def greet_user(self):
        print(f"Приветствую вас {self.first_name} {self.last_name}")
   
    def increment_login_attempts(self):
        self.login_attempts += 1
    
    def reset_login_attempts(self):
        self.login_attempts = 0
    
user = User("Kairat", "Omarov")
   
user.increment_login_attempts()
user.increment_login_attempts()
user.increment_login_attempts()
user.increment_login_attempts()
user.increment_login_attempts()

print(user.login_attempts)
user.reset_login_attempts()
print(user.login_attempts)
```

```python
from random import randint
class Die():
    def __init__(self, sides):
        self.sides = sides

        
    def roll_die(self):
        num = randint(1, self.sides)
        print(f"Результат броска: {num}")
        
die = Die(6)
die.roll_die()
die.roll_die()
die.roll_die()
  
```

```python
die = Die(10)

die.roll_die()
die.roll_die()

die = Die(20)
die.roll_die()

die.roll_die()
  
```
