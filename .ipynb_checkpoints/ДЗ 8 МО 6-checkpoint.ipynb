```python
import pandas as pd
# Задание 1
# Чтение CSV-файла
# Основы чтения и записи данных:
df = pd.read_csv('Таблица для ДЗ.csv')
```

```python
# Задание 2 - Задачи на использование .loc:
# Выбрать продажи для определенной даты:
```

```python
# Продажи на конкретную дату
sales_on_date = df.loc[df["Date"] == "2020-04-15"]
```

```python
sales_on_date
```

```python
# Выбрать данные по диапазону индексов и определенным колонкам:
# Строки с 5 по 10 (включительно) и колонки Product, Sales
subset = df.loc[5:10, ["Product", "Sales"]]
```

```python
subset
```

```python
# Изменить данные для определенного условия:
# Изменение значения Returned, если рейтинг = 1
df.loc[df["Customer_Rating"] == 1, "Returned"] = True
print(df[df["Customer_Rating"] == 1])
```

```python
# Задание 3 - Задачи на группировку данных:
# Группировка по продукту и подсчет среднего возраста покупателей:
# Средний возраст по продуктам
avg_age_per_product = df.groupby("Product")["Customer_Age"].mean()
```

```python
avg_age_per_product
```

```python
# Сумма продаж по рейтингу
total_sales_by_rating = df.groupby("Customer_Rating")["Sales"].sum()
```

```python
total_sales_by_rating
```
