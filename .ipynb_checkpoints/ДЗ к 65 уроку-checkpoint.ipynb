```python
%matplotlib inline

import matplotlib as mpl
import matplotlib.pyplot as plt

import numpy as np
```

```python
#1
x = np.linspace(0, 30, 150)
print("Линейный массив: ")
x
```

```python
y = np.exp(x)
```

```python
plt.plot(x, y)
plt.title('График экспоненциальной функции')
plt.xlabel('x')
plt.ylabel('exp(x)')
plt.grid(True)
plt.show()
```

```python
#2
x = np.linspace(1, 10)
print("Линейный массив: ")
x
```

```python
y = np.log(x)
y1 = np.log2(x)
y2 = np.log10(x)

plt.plot(x, y, label='ln(x)')
plt.plot(x, y1, label='log2(x)')
plt.plot(x, y2, label='log10(x)')

plt.title('Графики логарифмических функций')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#3
x = np.linspace(0, 1)
print("Линейный массив: ")
x
```

```python
y = np.arcsin(x)
y1 = np.arccos(x)
y2 = np.arctan(x)
plt.plot(x, y, 'red', linestyle = "dashed")
plt.plot(x, y1, 'green', linestyle = 'dotted')
plt.plot(x, y2, 'yellow', linestyle = '-.')
 
plt.title('Графики арктангенсовых функций')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
    
plt.savefig('arctan_plot.png')
plt.show()
```

```python
#4
x = np.linspace(0, 10)
print("Линейный массив: ")
x
```

```python
y = np.sinh(x)
plt.plot(x, y)
   
plt.xlim(-2, 12)
plt.ylim(-1000, 12000)
plt.xlabel('Линейный массив')
plt.ylabel('sinh(x)')
plt.title('График гиперболического синуса')
plt.grid(True)
plt.show()
```

```python
#5
x2 = np.linspace(1, 10)
y2 = np.log(x2)
y21 = np.log2(x2)
y22 = np.log10(x2)
    
x3 = np.linspace(0, 1)
y3 = np.arcsin(x3)
y31 = np.arccos(x3)
y32 = np.arctan(x3)
    
fig, axs = plt.subplots(2, 3, figsize=(15, 10))
    
axs[0, 0].plot(x2, y2)
axs[0, 0].set_title('ln(x)')
   
axs[0, 1].plot(x2, y21)
axs[0, 1].set_title('log2(x)')

axs[0, 2].plot(x2, y22)
axs[0, 2].set_title('log10(x)')

axs[1, 0].plot(x3, y3, 'red', linestyle = 'dashed')
axs[1, 0].set_title('arcsin(x)')

axs[1, 1].plot(x3, y31, 'green', linestyle = 'dotted')
axs[1, 1].set_title('arccos(x)')
  
axs[1, 2].plot(x3, y32, 'yellow', linestyle = '-.')
axs[1, 2].set_title('arctan(x)')
    
plt.suptitle('Графики функций', fontsize=16)
plt.show()
```
