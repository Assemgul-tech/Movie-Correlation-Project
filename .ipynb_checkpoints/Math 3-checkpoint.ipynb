```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

```python
#1. Создайте функцию f(x) = 3 * x^2 + 2 * x - 10
```

```python
def f(x):
    return 3 * x**2 + 2 * x - 10
```

```python
#2. Создайте список x_points (или массив через numpy), в котором у нас лежат числа от -10 до 10
```

```python
def f_diff(x):
    return 6 * x + 2
```

```python
#3. Используйте x_points как аргумент для созданной функции f(x), и выведите на экран результаты функции для каждого значения 
# x_points, также сохраните результаты в y_points (cписок или массив)
```

```python
x_points = np.linspace(-10, 10, 400)
```

```python
#4. Покажите на графике x_points & y_points
```

```python
y_points = f(x_points)
```

```python
#5. Найдите производную функции f(x) и создайте новую функцию на ее основе как f_diff(x)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_points, y_points, label='f(x)', color='blue')
plt.title('График функции f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#6 Чему равна производная при x = 5
```

```python
derivative_at_5 = f_diff(5)
print(f"Производная в точке x=5 равна: {derivative_at_5}")
```

```python
#7. Покажите на графике f(x) & f_diff(x)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)', color='blue')
plt.plot(x_points, f_diff(x_points), label="f'(x)", color='green')
plt.title('График функции и её производной')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#8. Найдите уравнение касательной линии в точке x = 8
```

```python
x0 = 8
y0 = f(x0)
slope = f_diff(x0)
```

```python
def tangent_line(x):
    return slope * (x - x0) + y0
```

```python
#9. Покажите на графике функцию f(x) & касательную линию в точке x = 8
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)', color='blue')
plt.plot(x_points, tangent_line(x_points), label='Касательная в x=8', color='red', linestyle='--')
plt.scatter(x0, y0, color='black', zorder=5)  # Точка касания
plt.title('Функция и касательная в x=8')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#10. Найдите мин или макс функции f(x) и покажите на графике касательную линию, которая проходит через этот мин или макс функции
```

```python
a = 3
b = 2
vertex_x = -b / (2 * a)
vertex_y = f(vertex_x)
print(f"Минимум функции: x = {vertex_x:.2f}, y = {vertex_y:.2f}")

# Уравнение касательной в вершине
slope_vertex = f_diff(vertex_x)

def tangent_at_vertex(x):
    return slope_vertex * (x - vertex_x) + vertex_y

# График функции и касательной в вершине
plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)', color='blue')
plt.plot(x_points, tangent_at_vertex(x_points), label='Касательная в минимуме', color='orange', linestyle='--')
plt.scatter(vertex_x, vertex_y, color='black', zorder=5)
plt.title('Функция и касательная в минимуме')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#11. Посчитайте интеграл функции f(x) между 0 и 5. Также отобразите на графике площадь, которую мы посчитали 
```

```python
integral_value, _ = quad(f, 0, 5)
print(f"Интеграл функции от 0 до 5: {integral_value:.2f}")

# Отобразим площадь под кривой между 0 и 5
x_fill = np.linspace(0, 5, 100)
y_fill = f(x_fill)

plt.figure(figsize=(10, 6))
plt.plot(x_points, f(x_points), label='f(x)', color='blue')
plt.fill_between(x_fill, y_fill, alpha=0.3, color='cyan', label='Площадь между 0 и 5')
plt.title('Площадь под графиком функции')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
