```python
# Вопрос 1. Набор данных включает как отмененные, так и завершенные рейсы. Определите, есть ли больше 
# выполненных или отмененных рейсов, и посчитайте числовую разницу между ними.


# Количество выполненных и отменённых рейсов
completed = (df['Cancelled'] == 0).sum()
cancelled = (df['Cancelled'] == 1).sum()

print(f"Выполнено рейсов: {completed}")
print(f"Отменено рейсов: {cancelled}")
print(f"Разница: {abs(completed - cancelled)}")

# Вывод, кто больше
if completed > cancelled:
    print("Выполнено рейсов больше отменённых")
else:
    print("Отменённых рейсов больше выполненных")
```

```python
# Вопрос 2. Найдите рейс с наибольшей задержкой вылета и рейс с наибольшей задержкой прибытия. 
# Летят ли они в один аэропорт, и если да, то какой у него код?

columns = [
    'DepDelay', 'ArrDelay', 'UniqueCarrier', 'FlightNum', 'Dest'
]
df = pd.read_csv('flights_2008.csv.bz2', usecols=columns)

# Рейс с максимальной задержкой вылета
dep_max = df.loc[df['DepDelay'].idxmax()]

# Рейс с максимальной задержкой прибытия
arr_max = df.loc[df['ArrDelay'].idxmax()]

print(f"Наибольшая задержка вылета: {dep_max['DepDelay']} мин, Dest: {dep_max['Dest']}")
print(f"Наибольшая задержка прибытия: {arr_max['ArrDelay']} мин, Dest: {arr_max['Dest']}")

if dep_max['Dest'] == arr_max['Dest']:
    print(f"Да, оба летят в: {dep_max['Dest']}")
else:
    print("Нет, летят в разные аэропорты.")


```

```python
# Вопрос 3. Найдите перевозчика, у которого наибольшее количество отмененных рейсов.

# Убедимся, что нужные столбцы есть
columns = ['UniqueCarrier', 'Cancelled']
df = pd.read_csv('flights_2008.csv.bz2', usecols=columns)
# Отменённые рейсы
cancelled_flights = df[df['Cancelled'] == 1]

# Подсчёт отмен по каждому перевозчику
carrier_cancel_counts = cancelled_flights['UniqueCarrier'].value_counts()

# Наибольшее количество отмен
most_cancelled_carrier = carrier_cancel_counts.idxmax()
cancel_count = carrier_cancel_counts.max()

print(f"Перевозчик с наибольшим числом отменённых рейсов: {most_cancelled_carrier} ({cancel_count} рейсов)")
```

```python
df["DepTime"] = pd.to_numeric(df["DepTime"], errors='coerce')  # на всякий случай
df["DepHour"] = df["DepTime"] // 100
df["DepHour"].replace(to_replace=24, value=0, inplace=True)
df["DepHour"].describe()
```

```python
# Вопрос 4. Давайте рассмотрим время отправления и рассмотрим распределение по часам (столбец DepHour, который мы создали ранее). В какой час самый высокий процент рейсов?
total_by_hour = df['DepHour'].value_counts(normalize=True).sort_values(ascending=False)
total_by_hour.head()
```

```python
# Вопрос 5. Хорошо, теперь давайте рассмотрим распределение отмененных рейсов по времени. 
# В какой час наименьший процент отмененных рейсов?

# Кол-во рейсов по часам
total = df.groupby('DepHour').size()

# Кол-во отменённых по часам
cancelled = df[df['Cancelled'] == 1].groupby('DepHour').size()

# Доля отменённых по часам
cancelled_rate = (cancelled / total).sort_values()

print(" Часы с наименьшим процентом отменённых рейсов:")
print(cancelled_rate.head())
```

```python
# Вопрос 6. Есть ли какой-нибудь час, в который вообще не было отмененных рейсов? Проверить все, что относится.

# Сгруппируем все рейсы по часу и посчитаем количество отмененных рейсов
cancelled_by_hour = df[df['Cancelled'] == 1].groupby('DepHour').size()

# Все возможные часы в данных
all_hours = sorted(df['DepHour'].unique())

# Проверим, какие часы отсутствуют в списке отмен — значит, там 0 отмен
hours_with_no_cancellations = [hour for hour in all_hours if hour not in cancelled_by_hour.index]

print("Часы без отмен:", hours_with_no_cancellations)
```

```python
# Вопрос 7. Найдите самый загруженный час или, другими словами, час, когда количество вылетающих рейсов достигает максимума.

busy_hour = df[df['Cancelled'] == 0]['DepHour'].value_counts().sort_values(ascending=False)
print(" Самый загруженный час (по числу вылетов):")
print(busy_hour.head())
```

```python
# Вопрос 8. Поскольку мы знаем час отправления, было бы интересно изучить среднюю задержку для соответствующего 
# часа. Бывают ли случаи, когда самолеты в среднем вылетали раньше положенного? И если да, то в какое время это 
# произошло?
# Только завершённые рейсы
completed_flights = df[df['Cancelled'] == 0]

avg_delay_by_hour = df[df['Cancelled'] == 0].groupby('DepHour')['DepDelay'].mean()
early_departures = avg_delay_by_hour[avg_delay_by_hour < 0]
print(" Часы с вылетами раньше времени (средняя задержка < 0):")
print(early_departures)
```

```python
# Вопрос 9. Учитывая только выполненные перевозчиком рейсы, которые вы нашли в вопросе 3, 
# найдите распределение этих рейсов по часам. В какое время вылетает наибольшее количество самолетов?

completed_flights_by_hour = df[df['Cancelled'] == 0]['DepHour'].value_counts().sort_values(ascending=False)
print(" Распределение завершённых рейсов по часам:")
print(completed_flights_by_hour.head())
```

```python
# Вопрос 10. Найдите топ-10 перевозчиков по количеству выполненных рейсов (столбец UniqueCarrier)?
# Что из перечисленного ниже нет в вашем списке топ-10?
# Фильтруем только выполненные рейсы
completed = df[df['Cancelled'] == 0]

top_carriers = df[df['Cancelled'] == 0]['UniqueCarrier'].value_counts().head(10)
print(" Топ-10 перевозчиков по числу выполненных рейсов:")
print(top_carriers)
```

```python
# Вопрос 11. Постройте распределение причин отмены рейса (CancellationCode).

cancel_reasons = df[df['Cancelled'] == 1]['CancellationCode'].value_counts()
reason_map = {'A': 'Перевозчик', 'B': 'Погодные условия', 'C': 'Национальная воздушная система', 'D': 'Причины безопасности'}
print(" Распределение причин отмены рейсов:")
for code, count in cancel_reasons.items():
    print(f"{reason_map.get(code, code)}: {count}")
```

```python
# Вопрос 12. Какой маршрут наиболее частый по количеству рейсов?

df['Route'] = df['Origin'] + '-' + df['Dest']
top_route = df['Route'].value_counts().head(1)
print(" Самый частый маршрут:")
print(top_route)
```

```python
# Вопрос 13. . Найдите топ-5 задержанных маршрутов (посчитайте, сколько раз они задерживались при отправлении). 
# Из всех рейсов по этим 5 маршрутам посчитайте все рейсы, погодные условия которых способствовали задержке.

delayed_routes = df[df['DepDelay'] > 0]['Route'].value_counts().head(5)
print(" Топ-5 маршрутов с задержками:")
print(delayed_routes)

top_routes = delayed_routes.index
weather_delays = df[(df['Route'].isin(top_routes)) & (df['WeatherDelay'] > 0)]
print(" Количество рейсов с погодной задержкой на этих маршрутах:", weather_delays.shape[0])
```

```python
# Вопрос 14. Изучите почасовое распределение времени отправления. Выберите все верные утверждения:
#  Рейсы равномерно распределены во временном интервале [0-23].
# В период с 0 до 4 утра рейсов значительно меньше, чем с 19:00 до 20:00.

hour_counts = df[df['Cancelled'] == 0]['DepHour'].value_counts().sort_index()
print(" Распределение рейсов по часам:")
print(hour_counts)

print("\n В период с 0 до 4 утра:", hour_counts.loc[0:4].sum())
print(" В период с 19 до 20:", hour_counts.loc[19:20].sum())
```

```python
# Вопрос 15. Покажите, как количество рейсов меняется со временем (ежедневно/еженедельно/ежемесячно) и 
# интерпретируйте полученные результаты.

# Выберите все правильные утверждения:

# Количество рейсов в выходные дни меньше, чем в будние (рабочие дни).
# Наименьшее количество рейсов приходится на воскресенье.
# Зимой рейсов меньше, чем летом.
# Подсказка: ищите официальные метеорологические зимние месяцы для Северного полушария.


# Приводим типы колонок к целым числам
df['Year'] = df['Year'].astype(int)
df['Month'] = df['Month'].astype(int)
df['DayofMonth'] = df['DayofMonth'].astype(int)

# Переименуем колонку 'DayofMonth' во временный столбец 'Day', чтобы to_datetime понял
df = df.rename(columns={'DayofMonth': 'Day'})

# Создаём колонку с датой
df['Date'] = pd.to_datetime(df[['Year', 'Month', 'Day']])

# Возвращаем исходное имя колонки
df = df.rename(columns={'Day': 'DayofMonth'})

# Добавляем день недели (0 – понедельник, 6 – воскресенье)
df['Weekday'] = df['Date'].dt.weekday

# Проверим результат
print("Первые 5 строк с датой и днём недели:")
print(df[['Date', 'Weekday']].head())
```

```python
# Фильтруем только неотменённые рейсы
not_cancelled = df[df['Cancelled'] == 0]

# Ежедневное количество рейсов
daily_counts = not_cancelled.groupby('Date').size()

# По дням недели (0 – понедельник, 6 – воскресенье)
weekday_counts = not_cancelled.groupby('Weekday').size()

# По месяцам (1 – январь, 12 – декабрь)
monthly_counts = not_cancelled.groupby('Month').size()

# Вывод
print("\n Среднее количество рейсов в день (всего дней:", len(daily_counts), ")")
print("Минимум:", daily_counts.min(), "Максимум:", daily_counts.max(), "Среднее:", round(daily_counts.mean()))

print("\n Рейсы по дням недели (0 – понедельник, 6 – воскресенье):")
print(weekday_counts)

print("\n Рейсы по месяцам:")
print(monthly_counts)

```

```python
# Вопрос 16. Изучите распределение причин отмены во времени. Составьте диаграмму причин отмены, агрегированных по месяцам.

# Выберите все правильные утверждения:

# В октябре наименьшее количество отмен из-за погоды.
# Наибольшее количество отмен в сентябре связано с соображениями безопасности.
# Основная причина отмены рейсов в апреле – операторы связи.
# Отмены рейсов по вине Национальной воздушной системы происходят чаще, чем по вине перевозчиков.

cancel_by_month = df[df['Cancelled'] == 1].groupby(['Month', 'CancellationCode']).size().unstack(fill_value=0)
print(" Причины отмен по месяцам:")
print(cancel_by_month)
```

```python
# Вопрос 17. В каком месяце происходит наибольшее количество отмен по вине Carrier?

# Май
# Январь
# Сентябрь
# Апрель

carrier_cancel_month = df[df['CancellationCode'] == 'A'].groupby('Month').size().sort_values(ascending=False)
print(" Отмены по вине перевозчика по месяцам:")
print(carrier_cancel_month)
```
