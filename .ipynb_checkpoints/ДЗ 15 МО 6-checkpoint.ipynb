```python
import pandas as pd

# Загрузка данных
df = pd.read_csv('Student_bucketing.csv')
```

```python
# Просмотр первых строк
print(df.head())
```

```python
# Информация о столбцах и типах данных
print(df.info())
```

```python
# Описательная статистика
print(df.describe(include='all'))
```

```python
# Удалите колонку 'LikePresentations', она нам не понадобится
df = df.drop(columns=['LikePresentations'], errors='ignore')  # если вдруг её нет, не будет ошибки
```

```python
# Измените названия ваших колонок - разделите слова в названии колонок. Например, замените AcademicPerformance 
# на Academic Performance
df = df.rename(columns={
    'Student_id': 'Student ID',
    'Grade': 'Grade',
    'Employed': 'Employed',
    'marks': 'Marks'
})
```

```python
df.head
```

```python
# Очистка данных. Удалите дубликаты, если они есть и строки с пропущенными значениями
# Удалим дубликаты
df = df.drop_duplicates()
```

```python
df.shape
```

```python
# Удалим строки с пропущенными значениями
df = df.dropna()
```

```python
df.shape
```

```python
# Выявить среднее количество оценок студентов
excellent = df[df['Marks'] >= 90]
avg_marks_excellent = excellent['Marks'].mean()
print(f"Средний балл студентов с отличными оценками: {avg_marks_excellent:.2f}")
```

```python
 # посчитать среднее значение их баллов
employed_yes = df[df['Employed'].str.lower() == 'yes']
avg_marks = employed_yes['Marks'].mean()
print(f"Средний балл (marks) студентов с работой: {avg_marks:.2f}")
```
