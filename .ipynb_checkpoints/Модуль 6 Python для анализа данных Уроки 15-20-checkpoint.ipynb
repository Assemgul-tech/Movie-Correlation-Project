```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

iris = sns.load_dataset('iris')
```

```python
iris.describe()
```

```python
iris.shape
```

```python
iris.info()
```

```python
# Урок 18 Анализ датасета Iris с использованием Matplotlib в Python
```

```python
plt.figure(figsize=(10,6))
iris['sepal_length'].plot(kind='hist', bins=20, title='Sepal Lenght Distribution')
plt.show()
```

```python
iris['sepal_length'].plot
```

```python
# Урок 19 Анализ датасета Iris с использованием Matplotlib в Python
```

```python
plt.figure(figsize=(20,6))
iris.boxplot(column=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
plt.title('Iris Boxplot')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(iris['sepal_length'], iris['sepal_width'])
plt.title('Sepal Length vs Width')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
print(tips.head())
```

```python
tips.describe()
```

```python
tips['smoker'].unique()
```

```python
sns.histplot(tips['total_bill'], kde=True)
plt.title('Распределение общего счета')
plt.show()
```

```python
sns.histplot(tips['tip'], kde=True)
plt.title('Распределение общего счета')
plt.show()
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Boxplot общего счета по дням недели')
plt.show()
```

```python
sns.pairplot(tips)
plt.show()
```

```python
corr = tips.corr()
sns.heatmap(corr, annot=True)
plt.show()
```
