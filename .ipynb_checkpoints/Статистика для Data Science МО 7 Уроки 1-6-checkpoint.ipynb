```python
x_list = []
y_list = []

for i in range(10):
    x_list.append(i)
    y_list.append(2 * i + 1)

print(x_list)
print(y_list)
```

```python
import matplotlib.pyplot as plt

plt.plot(x_list, y_list, marker = 'o', linestyle='--', color = 'green', label= 'linear function')
plt.legend()
plt.show()
```

```python
x_list = []
y_list = []

for i in range(-10, 10):
    x_list.append(i)
    y_list.append(i ** 2 + 1)
    
print(x_list)
print(y_list)

plt.plot(x_list, y_list, marker ='o', linestyle='--', color = 'blue', label ='non-linear function')
plt.legend()
plt.show()
```

```python
x_list = []
y_list = []

for i in range (1, 10):
    x_list.append(i)
    y_list.append(i**5)
print(x_list)
print(y_list)

plt.plot(x_list, y_list, marker ='o', linestyle='--', color = 'green', label = 'Function')

y_list_derivative=[]

for i in range(1,10):
    y_list_derivative.append(5 *i**4)
print(y_list_derivative)

plt.plot(x_list, y_list_derivative, marker ='o', linestyle='--', color = 'red', label = 'Derivative')
plt.legend()
plt.show()
```

```python
x_tangent = []
y_tangent = []

for i in range(7,10):
    x_tangent.append(i)
    y_tangent.append(20480 * i -131072)
print(x_tangent)
print(y_tangent)

plt.plot(x_tangent,y_tangent, marker ='o', linestyle='--', color = 'blue', label = 'Tangent Line')
plt.legend()
plt.show()
```

```python
plt.plot(x_list, y_list, marker ='o', linestyle='--', color = 'green', label = 'Function')
plt.plot(x_tangent,y_tangent, marker ='o', linestyle='--', color = 'blue', label = 'Tangent Line')

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Создаем массив x от -10 до 10
x = np.linspace(-10, 10, 20)  # 20 точек

# 2. Создаем y = 0 для касательной (например просто горизонтальная линия на уровне y=0)
y_tangent = np.zeros_like(x)

# 3. Допустим, хотим показать просто точки x и y = 0
plt.plot(x, y_tangent, linestyle='-', color='blue', label='Касательная')

# 4. Дополнительно нарисуем сами точки x
plt.plot(x, y_tangent, linestyle='--', marker='o', color='red', label='Точки на касательной')

plt.title('Касательная линия и точки')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
