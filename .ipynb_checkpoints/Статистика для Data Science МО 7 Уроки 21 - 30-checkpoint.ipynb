```python
# Пример 1: равномерное распределение 
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(105)
data =np.random.uniform(0,100,10000).tolist()

ticks =np.arange(0,110,10).tolist()
std = round(np.std(data), 1)

plt.hist(data, bins=ticks, alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(np.mean(data), color='red', linestyle='dashed',linewidth=2, label=f'Mean')
plt.axvline(np.mean(data) + np.std(data), color='orange',linestyle='dashed',linewidth=2, label=f'Mean + SD({std})')
plt.axvline(np.mean(data) - np.std(data), color='orange',linestyle='dashed',linewidth=2, label=f'Mean - SD(-{std})')
plt.legend()

            
plt.title('Гистограмма с линией стандартного отколнения', fontsize=14)
plt.xlabel('Значения', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.legend(loc='lower left')            

plt.show()
```

```python
# Пример 2: уменьшим размер вдвое 
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(105)
data =np.random.uniform(0,50,10000).tolist()

ticks =np.arange(0,60,10).tolist()
std = round(np.std(data), 1)

plt.hist(data, bins=ticks, alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(np.mean(data), color='red', linestyle='dashed',linewidth=2, label=f'Mean')
plt.axvline(np.mean(data) + np.std(data), color='orange',linestyle='dashed',linewidth=2, label=f'Mean + SD({std})')
plt.axvline(np.mean(data) - np.std(data), color='orange',linestyle='dashed',linewidth=2, label=f'Mean - SD(-{std})')
plt.legend()

            
plt.title('Гистограмма с линией стандартного отколнения', fontsize=14)
plt.xlabel('Значения', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.legend(loc='lower left')            

plt.show()
```

```python
# Урок 24 Нормальное Распределение
```

```python
np.random.seed(105)
normal_data = np.random.normal(50, 15, 10000)
mean = np.mean(normal_data)
std = np.std(normal_data)
```

```python
print(mean, std)
```

```python
plt.hist(normal_data, rwidth=0.9, bins=np.arange(0, 105, 5),density=True)
plt.axvline(mean,linestyle='-', color='black',label='mean')
plt.axvline(mean+std,linestyle='--', color='red', label='+-1std')
plt.axvline(mean-std,linestyle='--', color='red')
plt.axvline(mean+std*2,linestyle='--', color='purple', label='+-2std')
plt.axvline(mean-std*2,linestyle='--', color='purple')
plt.axvline(mean+std*3,linestyle='--', color='green', label='+-3std')
plt.axvline(mean-std*3,linestyle='--', color='green')
plt.title('Гистограмма вероятности')
plt.ylabel('Вероятность')
plt.legend()
plt.show()
```

```python
# Урок 25 Оценка Плотности (KDE)
```

```python
sns.kdeplot(normal_data)
plt.hist(normal_data, rwidth=0.9, bins=np.arange(0,105,5), density=True)
plt.show()
```

```python
def normal_distribution_probability(x):
    power = -(x - mean)**2/(2*std**2)
    coef = 1/std/np.sqrt(2*np.pi)
    probability = coef* np.exp(power)
    return(probability)
```

```python
sns.kdeplot(normal_data)
plt.show()
```

```python
 normal_distribution_probability(40)
```

```python
# Урок 28 Стандартное нормальное распределение
```

```python
np.random.seed(105)
normal_sample = np.random.normal(5,2,1000)
```

```python
stdv = np.std(normal_sample)
mean = np.mean(normal_sample)
print(std, mean)
```

```python
np.random.seed(105)
standard_normal_sample = np.random.normal(0,1,1000)
```

```python
plt.hist(normal_sample,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='просто нормальное распределение')
plt.hist(standard_normal_sample,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='стандартное нормальное распределение')
plt.legend()
plt.show()
```

```python
normal_sample_shifted = normal_sample-mean
```

```python
plt.hist(normal_sample_shifted,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='просто нормальное распределение')
plt.hist(standard_normal_sample,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='стандартное нормальное распределение')
plt.legend()
plt.show()
```

```python
normal_sample_scaled = normal_sample_shifted / std
```

```python
plt.hist(normal_sample_scaled,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='просто нормальное распределение')
plt.hist(standard_normal_sample,rwidth=0.9, density=True, bins=np.arange(-5,10,0.5), alpha=0.5, label='стандартное нормальное распределение')
plt.legend()
plt.show()
```
