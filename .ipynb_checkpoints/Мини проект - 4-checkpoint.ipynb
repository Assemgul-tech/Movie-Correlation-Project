```python
# 2) Создайте класс Users со следующими атрибутами и функциями:
 a) имя
 b) фамилия
 c) год рождения
 d) адрес электронной почты
 e) getAge(): возвращает возраст пользователя, принимая во внимание, что текущий год 2023
```

```python
class Users:
    def __init__(self, first_name, last_name, birth_year, email):
        self.first_name = first_name    # Имя
        self.last_name = last_name      # Фамилия
        self.birth_year = birth_year    # Год рождения
        self.email = email              # Адрес электронной почты

    def getAge(self):
        return 2023 - self.birth_year

    def display_info(self):
        # Вывод информации об объекте
        print("Имя:", self.first_name)
        print("Фамилия:", self.last_name)
        print("Год рождения:", self.birth_year)
        print("Электронная почта:", self.email)
        print("Возраст:", self.getAge())

# Создаем объект и выводим его информацию:
user = Users("Magzhan", "Kairanbay", 1991, "magzhan.kairanbay@gmail.com")
user.display_info()
```

```python
# 3) преобразовать каждую строку файла в объект класса Users и сохранить их все в списке. Чтобы разделить слова,
разделенные запятой, используйте функцию split(). Например, если: s = "a,b,c", тогда t = s.split(',') 
вернет ['a', 'b', 'c']. Таким образом разделите данные в каждой строке на отдельные части и присвойте эти значения 
атрибутам класса «Users».
```

```python
class Users:
    def __init__(self, first_name, last_name, birth_year, email):
        self.first_name = first_name
        self.last_name = last_name
        self.birth_year = int(birth_year)
        self.email = email

    def getAge(self):
        return 2023 - self.birth_year

    def display_info(self):
        print("Имя:", self.first_name)
        print("Фамилия:", self.last_name)
        print("Год рождения:", self.birth_year)
        print("Email:", self.email)
        print("Возраст:", self.getAge())
        print("---------------------------------")

users_list = []

try:
    with open("users.txt", "r", encoding="utf-8") as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            parts = line.split(',')
            # Теперь проверяем, что строка содержит 5 элементов
            if len(parts) != 5:
                print("Неверный формат строки:", line)
                continue
            # Игнорируем первый элемент (ID) и используем остальные
            user_obj = Users(parts[1], parts[2], parts[4], parts[3])
            users_list.append(user_obj)
except FileNotFoundError:
    print("The file that you would like to read is not found")
except Exception:
    print("Some error occurred")

for user in users_list:
    user.display_info()
```

```python
# 4) написать функцию filter(low, high, users), которая будет принимать параметры по умолчанию "low"=30, 
"high"=50 и простой параметр - список объектов класса Users. Функция должна пройтись по всем пользователям и 
проверить возраст с параметрами low и high. Мы должны вернуть только тех пользователей, чей возраст выше нижнего 
параметра low и ниже верхнего параметра high. Распечатайте их адрес электронной почты и возраст. Пожалуйста,
разделите данные (адрес электронной почты и возраст) запятой. Затем вызовите эту функцию.
```

```python
def filter(low=30, high=50, users=[]):
    """
    Функция проходит по списку объектов Users и выводит адрес электронной почты и возраст
    для пользователей, чей возраст больше low и меньше high.
    
    Параметры:
      low (int): нижняя граница возраста (по умолчанию 30)
      high (int): верхняя граница возраста (по умолчанию 50)
      users (list): список объектов класса Users
      
    Возвращает:
      list: список пользователей, удовлетворяющих условию
    """
    filtered_users = []
    for user in users:
        age = user.getAge()
        if age > low and age < high:
            filtered_users.append(user)
            # Выводим адрес электронной почты и возраст, разделённые запятой
            print(f"{user.email}, {age}")
    return filtered_users

# Пример вызова функции, если список пользователей хранится в переменной users_list:
filtered = filter(users=users_list)
```
