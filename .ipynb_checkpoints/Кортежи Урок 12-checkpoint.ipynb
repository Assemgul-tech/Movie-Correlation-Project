# Множества

```python
# Урок 13 Множества через фигурные скобки создаются только уникальнын значения
```

```python
my_set = {1, 2, 3, 4}
```

```python
my_set
```

```python
type(my_set)
```

```python
my_set = set([1, 2, 3, 4])
```

```python
type(my_set)
```

```python
dublicate_set = {1, 2, 2, 2, 3, 4, 4}
```

```python
dublicate_set #  уникальность элементов
```

```python
dublicate_set.add(6) # добавление элементов
```

```python
dublicate_set
```

```python
dublicate_set.remove(2) # выдает ошибку , если есть значение
```

```python
dublicate_set
```

```python
dublicate_set.discard(10) # не выдает ошибку, если нет ошибки
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

```python
A|B # Обьеденение
```

```python
A.union(B) # Обьеденение
```

```python
A & B # Пересечение
```

```python
A.intersection(B)  # Пересечение
```

```python
A - B # Разница
```

```python
B - A # Разница
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6} # Cимметричная разница
result = A.symmetric_difference(B)
print(result)
```

```python
 B.symmetric_difference(A)
```

```python
A = {1, 2, 3, 4, 5}
```

```python
B = {3, 4}
```

```python
A.issubset(B) # Проверяет является ли  А множеством В
```

```python
B.issubset(A) # Проверяет является ли  А подмножеством В
```

```python
A.issuperset(B)
```
