```python
#5
import pandas as  pd

soccer_clubs = {'Роналду': 'Реал', 'Месси': 'Барселона', 'Суарес': 'Барселона', 'Гризманн ': 'Атлетико', 'Неймар': 'Барселона'}
clubs_series = pd.Series(soccer_clubs)
    
soccer_positions = {'Роналду': 'Нападающий', 'Месси': 'Нападающий', 'Суарес': 'Нападающий', 'Гризманн ': 'Нападающий', 'Неймар': 'Нападающий'}

positions_series = pd.Series(soccer_positions)
soccer_df = pd.DataFrame({'Club': clubs_series, 'Position': positions_series})
print("Исходный DataFrame:")
print(soccer_df)
   
soccer_df.at['Роналду', 'Position'] = 'Вратарь'
soccer_df.at['Месси', 'Position'] = 'Защитник'

print("Измененный DataFrame:")
print(soccer_df)  
```

```python
#6
import pandas as pd
import numpy as np
   
data = pd.DataFrame([[5, np.nan, 1], [4, 1, 2],[None, 3, 3]])

print("Исходный объект Series:")
print(data)
    
```

```python
print("Пустые значения (NA):")
data.isnull()
```

```python
print("Непустые значения:")
data[data.notnull()]
```

```python
print("Отброшены NA - значения:")
data.dropna()
```

```python
data = data.fillna(0)
print("Заполнены NA - значения:")
print(data)
```

```python
#7
data = pd.DataFrame({'Ferst': [1, np.nan, 4, 3], 'Second': [2, None, 1, 4]})
print("Исходный DataFrame:")
print(data)
```

```python
print("Отброшены строки с пустыми значениями:")
print(data.dropna(axis=0))
```

```python
print("Отброшены столбцы с пустыми значениями:")
print(data.dropna(axis=1))
```

```python
print("Заполнены строки по направлению вперед:")
print(data.fillna(method='ffill', axis=1))
 
```

```python
print("Заполнены строки по направлению назад:")
print(data.fillna(method='bfill', axis=1))
```

```python
print("Заполнены столбцы по направлению вперед:")
print(data.fillna(method='ffill', axis=0))
```

```python
print("Заполнены столбцы по направлению назад:")
print(data.fillna(method='bfill', axis=0))
 
```
