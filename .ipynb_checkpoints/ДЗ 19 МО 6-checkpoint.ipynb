```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Загрузка данных
df = pd.read_csv('coffee_sales_data.csv')

# Приводим названия столбцов к стандартному виду
df.columns = df.columns.str.strip()
```

```python
# Задание 1: Анализ общего количества продаж по дням недели
# Постройте столбчатую диаграмму, показывающую распределение общего количества продаж кофе по дням недели. 
# Используйте дни недели на оси X и общее количество продаж на оси Y. Какие дни недели имеют наибольшие продажи?
# Группировка по дню недели

sales_by_day = df.groupby('День недели')['Количество Продаж'].sum()

# Визуализация
plt.figure(figsize=(10, 5))
sns.barplot(x=sales_by_day.index, y=sales_by_day.values, palette='viridis')
plt.title('Общее количество продаж по дням недели')
plt.xlabel('День недели')
plt.ylabel('Количество продаж')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Задание 2: Анализ цен на разные типы кофе
# Создайте boxplot, который покажет распределение цен на разные типы кофе (Эспрессо, Латте, Капучино).
# Используйте тип кофе на оси X и цены на оси Y. Какой тип кофе имеет наибольший разброс цен?

plt.figure(figsize=(8, 5))
sns.boxplot(x='Тип Кофе', y='Цена за Чашку', data=df, palette='Set2')
plt.title('Распределение цен по типам кофе')
plt.xlabel('Тип кофе')
plt.ylabel('Цена за чашку')
plt.tight_layout()
plt.show()
```

```python
# Задание 3: Анализ взаимосвязи между выходными и продажами
# Исследовать влияние выходных дней на продажи кофе в магазине с помощью столбчатой диаграммы, 
# показывающей разницу в общем количестве продаж между выходными и будними днями. Определить, 
# в какие дни наблюдается больше продаж

# Убедимся, что столбец 'Выходной' в нужном формате
df['Выходной'] = df['Выходной'].str.strip().str.lower().map({'да': 'Выходной', 'нет': 'Будний день'})

# Группировка по признаку выходного
weekend_sales = df.groupby('Выходной')['Количество Продаж'].sum()

# Визуализация
plt.figure(figsize=(6, 5))
sns.barplot(x=weekend_sales.index, y=weekend_sales.values, palette='pastel')
plt.title('Сравнение продаж в выходные и будние')
plt.xlabel('День')
plt.ylabel('Количество продаж')
plt.tight_layout()
plt.show()
```
