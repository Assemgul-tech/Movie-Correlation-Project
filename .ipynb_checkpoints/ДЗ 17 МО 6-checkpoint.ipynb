```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# ЗАДАНИЕ 1: Линейная диаграмма температуры за неделю
days = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']
temperatures = [12, 14, 15, 13, 16, 18, 17]

plt.figure(figsize=(8, 4))
plt.plot(days, temperatures, marker='o', color='teal')
plt.title('Температура за неделю')
plt.xlabel('День недели')
plt.ylabel('Температура (°C)')
plt.grid(True)
plt.show()
```

```python
# ЗАДАНИЕ 2: Точечная диаграмма - часы учебы и оценки
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
grades = [50, 55, 60, 65, 70, 75, 85, 90]

plt.figure(figsize=(8, 4))
plt.scatter(study_hours, grades, color='purple')
plt.title('Зависимость оценки от часов учебы')
plt.xlabel('Часы учебы')
plt.ylabel('Оценки')
plt.grid(True)
plt.show()
```

```python
# ЗАДАНИЕ 3: Столбчатая диаграмма участников по видам спорта
sports = ['Футбол', 'Баскетбол', 'Волейбол', 'Плавание', 'Теннис']
participants = [25, 20, 18, 15, 10]

plt.figure(figsize=(8, 4))
plt.bar(sports, participants, color='coral')
plt.title('Участники по видам спорта')
plt.xlabel('Вид спорта')
plt.ylabel('Количество участников')
plt.show()
```

```python
# ЗАДАНИЕ 4: Гистограмма оценок
np.random.seed(0)  # Для повторяемости
grades = np.random.randint(2, 6, 100)  # 100 случайных оценок от 2 до 5

plt.figure(figsize=(8, 4))
plt.hist(grades, bins=8, color='skyblue', edgecolor='black')
plt.title('Распределение оценок')
plt.xlabel('Оценка')
plt.ylabel('Частота')
plt.show()
```

```python
# ЗАДАНИЕ 5: Две столбчатые диаграммы на одном изображении
coffee_types = ['Эспрессо', 'Латте', 'Капучино']
coffee_sales = [30, 45, 25]

snacks = ['Круассаны', 'Багеты', 'Печенье', 'Маффины', 'Пирожные']
snack_sales = [20, 15, 30, 18, 12]

fig, axs = plt.subplots(1, 2, figsize=(12, 5))
# Первая диаграмма
axs[0].bar(coffee_types, coffee_sales, color='saddlebrown')
axs[0].set_title('Продажи кофе за день')
axs[0].set_xlabel('Тип кофе')
axs[0].set_ylabel('Чашек продано')

# Вторая диаграмма
axs[1].bar(snacks, snack_sales, color='darkorange')
axs[1].set_title('Продажи закусок за день')
axs[1].set_xlabel('Закуска')
axs[1].set_ylabel('Штук продано')

plt.tight_layout()
plt.show()
```
