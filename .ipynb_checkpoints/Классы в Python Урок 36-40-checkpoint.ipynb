```python
class DataAnalyzer:
    def __init__(self, data):
        self.data = data # Атрибут класса
        
    def average(self):
        return sum(self.data)/ len(self.data) # Метод класса
```

```python
dataset = [1, 2, 3, 4, 5]
analyzer = DataAnalyzer(dataset)
print(analyzer.average())
```
