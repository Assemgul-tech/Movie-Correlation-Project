```python
country = ['France','Portugal','Italy','United Kingdom','Netherlands']
print(country)

country.sort()
print(country)
```

```python
country = ['France','Portugal','Italy','United Kingdom','Netherlands']
print(country)

country.sort(reverse=True)
print(country)
```

```python
age = 30
if age < 2:
    print("Baby")
elif age >= 2 and age < 4:
    print("Toddler")
elif age >= 4 and age < 13:
    print("Child")
elif age >= 13 and age < 20:
    print("Teen")
elif age >= 20:
    print("Adult")
```
