```python
# Задание 1 - Исследование нормальных распределений
- Сгенерируйте первый список из 1000 случайных чисел, следуя стандартному нормальному распределению (μ=0, σ=1). Сгенерируйте второй список из 1000 случайных чисел, следуя нормальному распределению с средним значением μ=8 и стандартным отклонением σ=2.
- Постройте гистограммы для обоих списков. Гистограммы должны быть нормализованы, чтобы отражать плотность вероятности.
- Оцентрируйте и измените масштаб нормального распределения с μ=8 и σ=2 так, чтобы его среднее значение стало 0, а стандартное отклонение 1. Постройте гистограмму для модифицированных данных, сравнивая её с гистограммой стандартного нормального распределения.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Генерация данных
np.random.seed(0)
data_std = np.random.normal(0, 1, 1000)
data_shifted = np.random.normal(8, 2, 1000)

# Стандартизация второго набора

data_standardized = (data_shifted - np.mean(data_shifted)) / np.std(data_shifted)

# Построение графиков
plt.figure(figsize=(12, 6))

# Гистограммы оригинальных распределений
plt.subplot(1, 2, 1)
plt.hist(data_std, bins=30, density=True, alpha=0.6, label='N(0,1)', color='blue')
plt.hist(data_shifted, bins=30, density=True, alpha=0.6, label='N(8,2)', color='orange')
plt.title('Гистограммы: N(0,1) и N(8,2)')
plt.xlabel('Значения')
plt.ylabel('Плотность')
plt.legend()

# Гистограммы стандартизированных данных
plt.subplot(1, 2, 2)
plt.hist(data_standardized, bins=30, density=True, alpha=0.6, label='Стандартизированные N(8,2)', color='green')
plt.hist(data_std, bins=30, density=True, alpha=0.6, label='N(0,1)', color='blue')
plt.title('Сравнение с N(0,1)')
plt.xlabel('Значения')
plt.ylabel('Плотность')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Задание 2 - Изучение распределения роста учащихся средней школы и его стандартизация для анализа.
- Используйте Numpy для создания выборки из 1000 случайных значений роста учащихся, следующих нормальному распределению с средним значением 170 см и стандартным отклонением 10 см.
- Преобразуйте исходные данные так, чтобы получить распределение с средним значением 0 и стандартным отклонением 1, используя простую стандартизацию.
- Постройте гистограммы для исходных данных роста учащихся и для стандартизированных данных.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Генерация роста: N(170, 10)
np.random.seed(42)
heights = np.random.normal(170, 10, 1000)

# Стандартизация (z-score): (x - mean) / std
heights_standardized = (heights - np.mean(heights)) / np.std(heights)

# Визуализация
plt.figure(figsize=(12, 5))

# Гистограмма исходных данных
plt.subplot(1, 2, 1)
plt.hist(heights, bins=30, color='skyblue', edgecolor='black', density=True)
plt.title('Рост учащихся (см)')
plt.xlabel('Рост (см)')
plt.ylabel('Плотность')

# Гистограмма стандартизированных данных
plt.subplot(1, 2, 2)
plt.hist(heights_standardized, bins=30, color='lightgreen', edgecolor='black', density=True)
plt.title('Стандартизованный рост')
plt.xlabel('Z-оценка')
plt.ylabel('Плотность')

plt.tight_layout()
plt.show()
```

```python
# Задание 3 - Изменение климата приводит к тому, что погодные условия становятся более переменчивыми, что включает изменения в среднесуточных температурах. В этом задании вы будете работать с моделью данных среднесуточных температур для определенного месяца, например, июля. Исходя из исторических данных, предположим, что средняя температура в июле составляет 22°C со стандартным отклонением 3°C.
- Сгенерируйте выборку из 1000 случайных значений среднесуточных температур для июля, используя нормальное распределение с заданными параметрами (среднее = 22°C, стандартное отклонение = 3°C).
- Определите, какой процент дней характеризуется аномально высокими (выше 27°C) и аномально низкими (ниже 15°C) температурами.
- Постройте гистограмму среднесуточных температур для июля. Выделите области аномальных температур на гистограмме разными цветами.
```

```python
# Генерация температур: N(22, 3)
temps = np.random.normal(22, 3, 1000)

# Расчет процентов аномальных температур
high_anomaly = np.sum(temps > 27)
low_anomaly = np.sum(temps < 15)
high_anomaly_percent = high_anomaly / len(temps) * 100
low_anomaly_percent = low_anomaly / len(temps) * 100

print(f"Аномально высоких дней (>27°C): {high_anomaly} ({high_anomaly_percent:.2f}%)")
print(f"Аномально низких дней (<15°C): {low_anomaly} ({low_anomaly_percent:.2f}%)")

# Визуализация
plt.figure(figsize=(10, 5))
bins = np.linspace(min(temps), max(temps), 40)

# Гистограмма всех температур
plt.hist(temps, bins=bins, color='lightgrey', edgecolor='black', label='Нормальные дни')

# Аномально высокие
plt.hist(temps[temps > 27], bins=bins, color='red', edgecolor='black', label='> 27°C')
# Аномально низкие
plt.hist(temps[temps < 15], bins=bins, color='blue', edgecolor='black', label='< 15°C')

plt.title('Распределение температур в июле')
plt.xlabel('Температура (°C)')
plt.ylabel('Количество дней')
plt.legend()
plt.tight_layout()
plt.show()
```
