```python
# 1. Расчет Суммы Чисел: Написать программу, которая использует цикл 
 for для вычисления суммы всех чисел от 1 до 100.
```

```python
sum_numbers = sum(range(1, 101))
print(f"Сумма чисел от 1 до 100: {sum_numbers}")
```

```python
# 2. Таблица Умножения: Используя цикл for, создать программу, которая выводит таблицу умножения 
 для выбранного пользователем числа от 1 до 10. Например, если пользователь выберет число 5, программа должна
вывести таблицу умножения на 5, то есть вот такой вывод:
```

```python
num = int(input("\nВведите число для таблицы умножения: "))
print(f"Таблица умножения для {num}:")
for i in range(1, 11):
    print(f"{num} * {i} = {num * i}")
```

```python
# 3. Генерация Списка Четных Чисел: Написать программу, использующую цикл for,
 для создания и вывода списка всех четных чисел от 2 до 20.
```

```python
even_numbers = [i for i in range(2, 21, 2)]
print("\nСписок четных чисел от 2 до 20:", even_numbers)
```

```python
#4. Обратный Отсчет: Создать программу, которая использует цикл for для вывода обратного отсчета 
    от 10 до 0, после чего выводит сообщение "Пуск!".
```

```python
print("\nОбратный отсчет:")
for i in range(10, -1, -1):
    print(i)
print("Пуск!")
```

```python
#5. Работа со Строками: Написать программу, которая принимает строку от пользователя и использует
 цикл for для вывода каждого символа этой строки в столбик. Например, для строки "Привет"
```

```python
text = input("\nВведите строку: ")
print("Символы в столбик:")
for char in text:
    print(char)
```
