```python
# Урок 15 True/False & IF
```

```python
x = 1
if x  > 3:
    x = x ** 2
    print(x)
    print("Hello world")
else:
    x = x/2
    print("Oops")
        
```

```python
x = 5

if x > 10:
    print("x больше 10")
elif x > 5:
    print("x больше 5, но не больше 10")
else:
    print("x не больше 5")
```

```python
x = 75
```

```python
if x > 100:
    print('У тебя легкая форма ожирения')
elif x > 80:
    print('У тебя все ок')
elif x > 60:
    print('Тебе надо задуматься над едой')
else:
    print('Я тебя не понимаю')
```

```python
x = 5
y = 7

if x == 5:
    if y == 10:
        print('x равно 5 и y равно 10')
    else:
        print('x равно 5, но y не равно 10')
```

```python
# and  и 
x = 5

if x > 3 and x < 10:
    print('x больше 3 и меньше 10')
```

```python
# or или
x = 5

if x > 3 or x < 10:
    print('x больше 3 и меньше 10')
```

```python
x = 5

if not x > 5:
    print('x больше 3 и меньше 10')
```
