```python
x = np.array([1, 2, 4, 6, 4, 3, 8])
x1, x2, x3 = np.split(x, [2, 4])

print(x1)
print(x2)
print(x3)
```

```python
kvadrat = np.arange(16).reshape(4, 4)
print(kvadrat)
```

```python
y1, y2 = np.vsplit(kvadrat, [2])

print(y1)
print()
print(y2)
```

```python
y3, y4 = np.hsplit(kvadrat,[2])

print(y3)
print()
print(y4)
```
