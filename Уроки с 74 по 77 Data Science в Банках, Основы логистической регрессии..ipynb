```python
df = pd.read_csv("banking.csv")
df.head()
```

```python
sdf = df.loc[:,['job', 'marital', 'education', 'default', 'housing', 'loan', 'y']]
sdf.head()
```

```python
sdf['marital'].unique()
```

```python
sdf['education'] = np.where(sdf['education'] == 'basic.4y', 'basic.school',sdf['education'])
sdf['education'] = np.where(sdf['education'] == 'basic.6y', 'basic.school',sdf['education'])
sdf['education'] = np.where(sdf['education'] == 'basic.9y', 'basic.school',sdf['education'])
sdf['education'].unique()
```

```python
sdf['default'].unique(),sdf['housing'].unique(),sdf['loan'].unique()
```

```python
# 77 урок Основы логистической регрессии
```

```python
data = pd.get_dummies(sdf, columns = ['job', 'marital', 'education', 'default', 'housing', 'loan'])
data.head()
```

```python
print(list(data.columns))
```

```python
data.drop(data.columns[[12, 16, 22, 24, 27, 30]], axis = 1, inplace = True)
```

```python
print(list(data.columns))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

x = data.drop('y', axis=1).values
y = data['y']
```

```python
x, y
```

```python
np.mean(y)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size = 0.8, stratify=y)
```

```python
np.mean(ytrain),np.mean(ytest)
```

```python
model = LogisticRegression()
model.fit(xtrain, ytrain)
```

```python
y_model = model.predict(xtest)

from sklearn.metrics import accuracy_score
accuracy_score(ytest, y_model)
```

```python
from sklearn.model_selection import cross_val_score

score = cross_val_score(model, x, y, cv=5)
score.mean()
```
