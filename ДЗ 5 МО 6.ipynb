```python
import numpy as np

# Задача 1: Сложение и изменение формы массивов
a1 = np.random.randint(1, 11, size=10)
a2 = np.random.randint(1, 11, size=10)
sum_array = a1 + a2
reshaped_array = sum_array.reshape(2, 5)
mean_value = reshaped_array.mean()

print("Задача 1:")
print("Массив 1:", a1)
print("Массив 2:", a2)
print("Сумма:", sum_array)
print("Форма 2x5:\n", reshaped_array)
print("Среднее значение:", mean_value)
```

```python
# Задача 2: Работа со срезами и формами
array_3x3 = np.random.randint(1, 11, size=(3, 3))
print("\nЗадача 2:")
print("Массив 3x3:\n", array_3x3)
print("Первая строка:", array_3x3[0])
print("Первый столбец:", array_3x3[:, 0])

reshaped = array_3x3.reshape(1, 9)
print("Изменённая форма (1x9):", reshaped)

flattened = array_3x3.flatten()
print("Преобразован в 1D:", flattened)
```

```python
# Задача 5: Операции над массивами
arr1 = np.random.randint(1, 6, size=(2, 2))
arr2 = np.random.randint(1, 6, size=(2, 2))
arr2_transposed = arr2.T
sum_result = arr1 + arr2_transposed

print("\nЗадача 5:")
print("Массив 1:\n", arr1)
print("Массив 2 (исходный):\n", arr2)
print("Массив 2 (транспонированный):\n", arr2_transposed)
print("Сумма массивов:\n", sum_result)
```
