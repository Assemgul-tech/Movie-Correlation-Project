```python
import pandas as pd

data = pd.DataFrame({
    'Shop':['Shop1','Shop1','Shop2','Shop2','Shop3'],
    'Product':['Apples','Oranges','Apples','Oranges','Apples'],
    'Sales':[10, 15, 5, 7, 6],
    'Quantity':[1, 3, 5, 7, 10]
})
```

```python
data
```

```python
data.groupby('Shop').agg([min, max, sum, 'mean'])
```

```python
numeric_cols = data.select_dtypes(include='number').columns
result = data.groupby('Shop')[numeric_cols].agg(['min', 'max', 'sum', 'mean'])
print(result)
```
