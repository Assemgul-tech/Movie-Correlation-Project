```python
# Задание 1: Словари
film = {
    "название": "Аватар",
    "год выпуска": 2009,
    "жанр": "Фантастика, боевик, драма, приключения",
    "режиссер": "Джеймс Кэмерон",
    "актеры": ["Сэм Уортингтон", "Зои Салдана", "Сигурни Уивер","Стивен Лэнг"]
}

print("\nИнформация о фильме:")
print(f"Название: {film['название']}")
print(f"Год выпуска: {film['год выпуска']}")
print(f"Жанр: {film['жанр']}")
print(f"Режиссер: {film['режиссер']}")
print(f"Актеры: {', '.join(film['актеры'])}")
```

```python
# Задание 2: Кортежи
data_tuple = (42, "Python", True, 3.14)

print("\nЭлементы кортежа с их типами:")
for item in data_tuple:
    print(f"{item} - {type(item)}")
```

```python
# Задание 3: Условный оператор IF
number = int(input("\nВведите число: "))

if number % 2 == 0:
    print("Число четное.")
else:
    print("Число нечетное.")
```
