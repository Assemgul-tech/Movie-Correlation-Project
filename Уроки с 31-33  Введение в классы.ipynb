```python
class Dog():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print(f"Собачка по имени {self.name} сейчас сидит")

    def roll_over(self):
        print(f"Собачка по имени {self.name} сейчас перекатывается")
        
        
my_dog = Dog("Barsik", 2)    
print(f"Мою собачку зовут {my_dog.name}")
print(f"Моей собачке {my_dog.age} года")

my_dog.sit()
```

```python
fourth_dog = Dog("Harry",4)

fourth_dog.sit()
fourth_dog.roll_over()

print()

fourth_dog.setName("Potter")
fourth_dog.setAge(2)

fourth_dog.sit()
fourth_dog.roll_over()

fourth_dog.increment_age(2)
print(f"Моей собачке {fourth_dog.age} лет")
```
