```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://arxiv.org/list/cs/new'
respone = requests.get(url)
```

```python
soup =  BeautifulSoup(respone.text,'html.parser')
```

```python
articles = soup.find_all('div', class_='meta')
for article in articles:
    title = article.find('div', class_="list-title mathjax").text.replace('Title:', '').strip()
    authors = article.find('div', class_='list-authors').text.replace('Authors:', '').strip()
    print(f'Назание статьи: {title}')
    print(f'Авторы: {authors}\n')
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL страницы с курсами валют ЕЦБ 
url = 'https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html'

# Отправляем HTTР запрос на URL
response = requests.get(url)

# Разбираем HTML контент страницы URL
soup = BeautifulSoup(response.text, 'html.parser')

# Ищем таблицу с курсами валют
table = soup.find('table',{'class':'forextable'})

# Инициализируем список для хранения данных
data = []

# Проверяем найдена ли таблица
if table:
    # Итерируем по каждой строке в таблице, кроме заголовочной строки
    for row in table.find_all('tr')[1:]:
        cols = row.find_all('td')
        # Извлекаем текст из каждой ячейки в строке
        currency = cols[2].text.strip()
        rate = cols[2].text.strip()
        # Добавляем данные в список
        data.append([currency, rate])
        
    # Преоброзуем список в DataFrame
    df = pd.DataFrame(data, columns=['Currency','Rate'])
    print(df)
else:
    print('Таблица не найдена.Пожалуйста, проверьте имя класса и убедитесь , что структура сайта не изменилась')
```
