```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("real_estate.csv")
df.head()
```

```python
print(list(df.columns))
```

```python
print(df.shape)
```

```python
sdf = df.loc[:,['X2 house age', 'X3 distance to the nearest MRT station', 'X4 number of convenience stores', 'Y house price of unit area']]
sdf.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

x = sdf.drop('Y house price of unit area', axis = 1). values
y = sdf['Y house price of unit area']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
    
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
print('Coefficient', model.coef_)
```

```python
import pandas as pd

y_pred = model.predict(x_test)

predictions_df = pd.DataFrame({'actual': y_test, 'forecast': y_pred})
predictions_df['difference'] = predictions_df['actual'] - predictions_df['forecast']
print(predictions_df.head())
```

```python
from sklearn import metrics

print("Mean Squarred Error: ", metrics.mean_squared_error(y_test, y_pred))
```
