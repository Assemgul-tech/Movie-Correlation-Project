🧠 1. Основные типы данных (очень важно!)
Python имеет несколько базовых типов:
🔹 int — целые числа
Примеры: 5, 42, -9
🔹 float — дробные числа
Примеры: 3.14, 0.5, -2.7
🔹 str — строки
Примеры: "hello", 'Astana', "Assemgul"
🔹 bool — логический тип
True или False
🔹 NoneType — специальный тип
None означает «ничего», «пусто».

```python
# 🧩 2. Проверка типа: type()
x = 5
print(type(x))   # <class 'int'>
```

```python
# 🔄 3. Приведение типов
int("5")       # превращает строку в число
str(10)        # превращает число в строку
float(3)       # превращает int в float
```

Списки (list) и операции с ними
🧠 1. Что такое список?
Список (list) — это упорядоченная коллекция элементов.
🛠 2. Индексация (очень важно!)
Элемент	A	B	C	D
Индекс	0	1	2	3


```python
numbers = [10, 20, 30, 40]
names = ["Assemgul", "Asem", "Gul"]
mixed = [10, "Asem", True, 3.14]
```

```python
letters = ["A", "B", "C", "D"]
print(letters[0])   # A
print(letters[2])   # C
```

```python
# 🔄 3. Изменение элементов
nums = [1, 2, 3]
nums[1] = 100
print(nums)   # [1, 100, 3]
```

```python
# ➕ 4. Добавление элементов
nums.append(50)     # в конец
nums.insert(1, 99)  # вставить на позицию 1
```

```python
# ➖ 5. Удаление элементов
nums.remove(100)  # удалить по значению
nums.pop(2)       # удалить по индексу
```

```python
# 💡 6. Полезные методы списков

len(list)          # длина списка
list.sort()        # сортировка
list.reverse()     # обратный порядок
max(list), min(list)
```
