```python
def find_max_min(numbers, max_or_min):
    if max_or_min == 'max':
        return max(numbers)
    else:
         return min(numbers)
```

```python
find_max_min([1, 2, 3], 'min')
```

```python
def filter_greater_than(numbers, threshold):
    empty_list = []
    
    for i in numbers:
        if i > threshold:
            empty_list.append(i)
            
    return empty_list
```

```python
filter_greater_than([1, 2, 3, 4], 2)
```
