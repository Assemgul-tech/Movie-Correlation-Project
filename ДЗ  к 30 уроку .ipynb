```python
# 2.2

def make_album(artist_name, album_name, none = 'number of tracks'):
    album = {'artist name': artist_name, 'album name': album_name}
    if none:
        album['number of tracks'] = {'number of tracks':none}
        return album
    
       
album1 = make_album('Usher', 'Hard')
album2 = make_album('Akon', 'Freedom', 5)
print(album1)
print(album2)
```

```python
# 3.2

def send_messages(messages, sent_messages):
    while messages:
        current_message = messages.pop()
        print(current_message)
        sent_messages.append(current_message)
messages = ["hello", "wonderful", "world"]
sent_messages = []
send_messages(messages, sent_messages)
print("Original messages:", messages)
print("Sent messages:", sent_messages)
        
    
```

```python
# 3.3

def send_messages(messages, sent_messages):
    while messages:
        current_message = messages.pop()
        print(current_message)
        sent_messages.append(current_message)

messages = ["hello", "wonderful", "world"]
sent_messages = []
send_messages(messages[:], sent_messages)
print("Original messages:", messages)
print("Sent messages:", sent_messages)

    
```

```python
# 4
def make_sandwich(*components):
    print("Your sandwich contains:")
    for component in components:
        print("- " + component)

make_sandwich("cheese", "tomato")
make_sandwich("cheese", "potato", "orange", "apple")
make_sandwich("cheese", "banana")
    
```
