```python
try:
    # Код , который может вызвать исключение
    number = int(input("Введите число:"))
except ValueError:
    # Код , который выполнится ,если в блоке try возникнет ValueError
    print("Это не число!")
```

```python
number = int(input("Введите число:"))
```

```python
try:
    # Код , который может вызвать исключение
    number = int(input("Введите число:"))
except ValueError:
    # Обработка ValueError
    print("Это не число!")
except KeyboardInterrupt:
    # Обработка исключения. когда пользователь прерывает выполнение
    print("Вы прервали ввод!")
```

```python
try:
    number = int(input("Введите число:"))
    print(5/number)
except ZeroDivisionError:
    print('Вы делите на 0')
except ValueError:
    # Обработка ValueError
    print("Это не число!")
```

```python
try:
    number = int(input("Введите число:"))
except ValueError:
    # Обработка ValueError
    print("Это не число!")
else:
    print("Вы ввели число ", number)
```

```python
try:
    number = int(input("Введите число:"))
except ValueError:
    # Обработка ValueError
    print("Это не число!")
finally:
    print("Этот блок выполняется всегда")
```

```python
def read_file():
    while True:
        filename = input("Ввелите имя файла или 'exit' для выхода:")
        if filename == 'exit':
            print("Выход из программы.")
            break
        try:
            with open(filename, 'r') as file:
                print(file.read())
            break # Выход из цикла, если файл успешно прочитан
        except FileNotFoundError:
            print("Файл не найден. Попробуйте снова.")
```

```python
read_file()
```
