```python
%matplotlib inline

import matplotlib as mpl
import matplotlib.pyplot as plt

import numpy as np
```

```python
x = np.linspace(0, 10, 100)
x
```

```python
fig = plt.figure()
y = np.sin(x)
y1 = np.cos(x)
plt.plot(x, y)
plt.plot(x, y1)
```

```python
fig.savefig('sincos_fig.png')
```

```python
# Урок 54 Библиотека Matplotlib
fig, ax = plt.subplots(2)

ax[0].plot(x, y)
ax[1].plot(x, y1)
```

```python
# Урок 57 Библиотека Matplotlib
ax = plt.axes()
ax.plot(x, y, '-.g')
ax.set(xlim=(-2, 12), ylim=(-2, 2), xlabel='Одномерный массив', ylabel='sin(x)', title='График синуса')
```

```python
# Урок 58 Библиотека Matplotlib
ax1 = plt.axes()
ax2 = plt.axes([0.5, 0.5, 0.35, 0.35])

ax1.plot(x, y)
ax2.plot(x, x)
```

```python
for i in range(1, 7):
    plt.subplot(2, 3, i)
    plt.plot(x, y1)
```

```python
fig, ax = plt.subplots(2, 3)

ax[0][0].plot(x, y)
ax[0][1].plot(x, y1)
ax[0][2].plot(x, np.tan(x))

ax[1][0].plot(x, x)
ax[1][1].plot(x, x**2)
ax[1][2].plot(x, x**2)
```
