```python
# Перейти на сайт tengrinews.kz.
- Спарсить главные заголовки новостей.
- Описание задания:
- Используя библиотеку BeautifulSoup и requests, получите HTML-код главной страницы сайта.
- Найдите все заголовки новостей на главной странице.
- Распечатайте результат в формате:
- Заголовок 1
- Заголовок 2
...
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get('https://tengrinews.kz')
time.sleep(3)
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
headlines = soup.find_all('span', class_='main-news_super_item_title')

for headline in headlines:
print(headline.text.strip())

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get('https://tengrinews.kz')  # ← убрана ; в конце

time.sleep(3)
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
headlines = soup.find_all('span', class_='main-news_super_item_title')

for headline in headlines:
    print(headline.text.strip())  # ← добавлен отступ для блока цикла

driver.quit()
```
