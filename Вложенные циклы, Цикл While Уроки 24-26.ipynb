```python
x_coordinates = [1, 2, 3]
y_coordinates = ['a', 'b', 'c']

for x in x_coordinates:
    for y in y_coordinates:
        print(f"({x}, {y})")
```

```python
final_number = int(input('enter some number:'))

x_coordinates = [1, 2, 3, 4, 5, 6]
y_coordinates = [1, 2, 3, 4, 5, 6]

for i in x_coordinates:
    for j in y_coordinates:
        if (i+j) == final_number:
            print(f"({i}, {j})")
```

```python
numbers = [1, 6, 8, 100, 17, 18, 35, -6]
```

```python
target_number = int(input("enter some number:"))
```

```python
for i in numbers:
    for j in numbers:
        if i + j == target_number:
            print(i, j)
```

```python
count = 0
while count < 10:
    count = count + 1
    if count == 5:
        continue
    if count == 8:
        break
    print(count)
```
