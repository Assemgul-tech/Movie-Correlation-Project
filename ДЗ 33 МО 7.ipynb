```python
# Задача 1 - Вероятность распределения и уровень значимости
- В кафе утверждают, что каждая порция эспрессо содержит ровно 30 мл кофе. 
- Однако, на основании многолетних наблюдений, известно, что объем порции подчиняется нормальному распределению
- со средним значением 30 мл и стандартным отклонением 1 мл. Контролер качества хочет проверить, какова вероятность
- того, что случайно выбранная порция эспрессо будет содержать менее 29 мл кофе, что считается неприемлемым для 
- стандартов кафе. Используя пороговое значение для уровня значимости 0.05 определите соответствует ли реальная
- практика утверждениям кафе
```

```python
from scipy.stats import norm

mu = 30
sigma = 1
x = 29

z = (x - mu) / sigma  # Z-оценка
p = norm.cdf(z)       # P(X < 29)

print(f"Z = {z:.2f}, P(X < 29) = {p:.4f}")
```

```python
# Задача 2 - Вероятность распределения и уровень значимости
- Фармацевтическая компания разработала новое лекарство, предназначенное для снижения уровня холестерина. 
- По результатам предварительных исследований, утверждается, что среднее снижение уровня холестерина после 
- применения лекарства составляет 20 мг/дл. Однако, на основании многолетних наблюдений за эффективностью 
- аналогичных препаратов, известно, что снижение уровня холестерина подчиняется нормальному распределению с 
- средним значением 18 мг/дл и стандартным отклонением 4 мг/дл. Врач хочет проверить, какова вероятность того, 
- что применение нового лекарства приведет к снижению уровня холестерина на 22 мг/дл или больше, что будет 
- считаться значительным улучшением по сравнению со стандартными лекарствами. Используя пороговое значение 0.01 
- определите такое снижение холестерина случайность или такого уровня снижения не может быть.Проставьте галочку, 
- после просмотра урока (это важно для шкалы модуля)
```

```python
mu = 18
sigma = 4
x = 22

z = (x - mu) / sigma
p = 1 - norm.cdf(z)  # P(X ≥ 22)

print(f"Z = {z:.2f}, P(X ≥ 22) = {p:.4f}")
```

```python
# Задача 3 - Центральная предельная теорема
Создайте свой список с случайными числами, пусть значения будут из нормального распределения, равномерного распределения и экспоненциального распределения, как на уроке.
● посчитайте среднее значение
● создайте рандомную выборку из 300 значений, посчитайте ее среднее значение
● постройте гистограмму и qq.plot
```

```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

np.random.seed(0)

# Три распределения
norm_data = np.random.normal(50, 10, 1000)
uniform_data = np.random.uniform(0, 100, 1000)
exp_data = np.random.exponential(10, 1000)

# Функция для анализа
def analyze_distribution(data, title):
    sample = np.random.choice(data, size=300, replace=False)
    mean_sample = np.mean(sample)
    
    print(f"{title} — среднее всей выборки: {np.mean(data):.2f}, среднее подвыборки: {mean_sample:.2f}")
    
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.hist(sample, bins=30, color='lightblue', edgecolor='black')
    plt.title(f'{title}: гистограмма')

    plt.subplot(1, 2, 2)
    sm.qqplot(sample, line='r')
    plt.title(f'{title}: Q-Q plot')
    plt.tight_layout()
    plt.show()

analyze_distribution(norm_data, "Нормальное распределение")

```

```python
analyze_distribution(uniform_data, "Равномерное распределение")
```

```python
analyze_distribution(exp_data, "Экспоненциальное распределение")
```

```python
# Задание 4 - Центральная предельная теорема
- Исследование показало, что среднее время, которое жители определённого города тратят на дорогу до работы, 
- составляет 30 минут. Вы хотите проверить, насколько вероятно, что выборка из 50 случайных жителей этого 
- города будет иметь среднее время в пути до работы, близкое к этому значению. (стандартное отклонение выборки = 10).
```

```python
mu = 30
sigma = 10
n = 50
se = sigma / np.sqrt(n)

p1 = norm.cdf(31, loc=mu, scale=se)
p2 = norm.cdf(29, loc=mu, scale=se)
prob = p1 - p2

print(f"P(29 ≤ X̄ ≤ 31) = {prob:.4f}")
```
