```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import numpy as np
import pandas as pd
```

```python
# y = ax + b
x = 10 * np.random.rand(50)
x
```

```python
y = 3 * x + 5 * np.random.rand(50)
y
```

```python
plt.scatter(x, y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
x.shape
```

```python
x[:, np.newaxis].shape
```

```python
model.fit(x[:, np.newaxis], y[:, np.newaxis])
```

```python
xfit = np.linspace(0, 10, 100)
yfit = model.predict(xfit[:, np.newaxis])
```

```python
plt.scatter(x, y)
plt.plot(xfit, yfit)
```

```python
# y = ax + b
print("slope(a): ", model.coef_[0])
print("intercept(b): ", model.intercept_)
```

```python
df = pd.read_csv("WineQt.csv")
df.head()
```

```python
df.shape
```

```python
x = df.drop('quality', axis = 1)
y = df['quality']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, stratify=y)
```

```python
reg = LinearRegression()
reg.fit(xtrain, ytrain)
```

```python
coef_df = pd.DataFrame(reg.coef_, x.columns, columns=['Coefficient'])
coef_df
```

```python
y_pred = reg.predict(xtest)
```

```python
df = pd.DataFrame({'Act': ytest, 'Pred': y_pred})
df
```

```python
from sklearn import metrics

print("Mean Squarred Error: ", metrics.mean_squared_error(ytest, y_pred))
```

```python
from sklearn.datasets import make_blobs
```

```python
x, y_true = make_blobs(n_samples = 300, centers = 4)
```

```python
plt.scatter(x[:, 0], x[:, 1])
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters = 4)
kmeans.fit(x)
```

```python
y_kmeans = kmeans.predict(x)
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y_kmeans, cmap='viridis')
```
