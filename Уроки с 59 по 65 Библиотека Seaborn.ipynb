```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('classic')

import numpy as np
import pandas as pd

x = np.linspace(0, 10, 500)

y = np.sin(x)
y1 = np.cos(x)

plt.plot(x, y)
plt.plot(x, y1)
```

```python
import seaborn as sns
sns.set()

plt.plot(x, y)
plt.plot(x, y1)
```

```python
# Урок 60 Библиотека Seaborn
sns.set_style("whitegrid")

plt.plot(x, y)
plt.plot(x, y1)
```

```python
sns.set_style("whitegrid", {'grid.color': 'green'})

plt.plot(x, y)
plt.plot(x, y1)
```
