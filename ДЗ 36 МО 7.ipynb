```python
# Задача 1 - Вероятность распределения
- Рекламное агентство заявляет, что каждая печатная брошюра содержит ровно 20 страниц. Однако, из прошлых данных
- о качестве печати известно, что количество страниц в брошюре подчиняется нормальному распределению с средним 
- значением 20 страниц и стандартным отклонением 2 страницы. Менеджер по контролю качества хочет проверить, 
- какова вероятность того, что случайно выбранная брошюра будет содержать менее 18 страниц, что будет считаться 
- недопустимым для стандартов агентства.
```

```python
from scipy.stats import norm

mu = 20
sigma = 2
x = 18

z = (x - mu) / sigma
p = norm.cdf(z)

print(f'Z = {z:.2f}, P(X < 18) = {p:.4f}')
```

```python
# Задача 2 - Оценка доверительного интервала
- Вы исследуете средний рост студентов университета. Вы измерили рост случайно выбранных студентов и хотите оценить,
- насколько реальное распределение роста соответствует вашим наблюдениям. Вы предполагаете, что рост студентов
- нормально распределен с средним значением 175 см и стандартным отклонением 6 см. Оцените 95% доверительный 
- интервал для среднего роста студентов и визуализируйте распределение с помощью KDE-графика.
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

mu = 175
sigma = 6
n = 100

sample = np.random.normal(mu, sigma, n)
sample_mean = np.mean(sample)
se = sigma / np.sqrt(n)
z = 1.96

ci_lower = sample_mean - z * se
ci_upper = sample_mean + z * se

print(f'95% доверительный интервал: ({ci_lower:.2f} см, {ci_upper:.2f} см)')

# Визуализация
sns.kdeplot(sample, fill=True, color='skyblue')
plt.axvline(ci_lower, color='red', linestyle='--', label='Нижняя граница CI')
plt.axvline(ci_upper, color='green', linestyle='--', label='Верхняя граница CI')
plt.axvline(sample_mean, color='blue', linestyle='-', label='Среднее значение')
plt.title('KDE-график роста студентов с 95% доверительным интервалом')
plt.xlabel('Рост (см)')
plt.legend()
plt.show()
```

```python
# Задача 3 - Точечная оценка
- Нужно смоделировать распределение оценок студентов по определенному предмету. Из предыдущих лет известно, что 
- средняя оценка по этому предмету составляет 75 баллов с стандартным отклонением 10 баллов. Мы хотим сгенерировать
- выборку из 100 оценок, чтобы проанализировать, как могут распределяться оценки текущего года. Посчитайте точечную 
- оценку.
```

```python
np.random.seed(0)
scores = np.random.normal(75, 10, 100)
mean_score = np.mean(scores)

print(f'Точечная оценка среднего балла: {mean_score:.2f}')
```
