```python
class DataAnalyzer:
    pi_value = 3.14
    
    def __init__(self, data):
        self.data = data # Атрибут класса
        
    def average(self):
        return sum(self.data)/ len(self.data) # Метод класса
    
    def sum_minus_5(self):
        return sum(self.data) - 5
```

```python
dataset = [1, 2, 3, 4, 5]
analyzer = DataAnalyzer(dataset)
print(analyzer.average())
```

```python
analyzer2 = DataAnalyzer([1, 2, 3])
```

```python
analyzer2.pi_value
```

```python
analyzer2.average()
```

```python
analyzer.data
```

```python
analyzer.sum_minus_5()
```

```python
class Cake:
    
    size = 20
    
    def __init__(self, flavor):
        self.flavor = flavor # Здесь мы говорим о вкусе (flavor) этого конкретного торта (self) будет тем6 что
        # мы передаем при создании.
        
    def describe(self):
        # А здесь мы говорим : расскажи  о вкусе (flavor) этого конкретного торта (self).
        print("Этот торт имеет вкус:", self.flavor)
```

```python
# Создаем торт со вкусом шоколоада
chocolate_cake = Cake("шоколадный")
chocolate_cake.describe() # Этот торт имеет вкус: шоколадный
```

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        
    def __str__(self):
        return f"Книга '{self.title}' автора {self.author}, {self.pages} стр."
    
    def is_long(self):
        return self.pages > 500
```

```python
# Создаем экземпляр класса Book
my_book = Book("Войнa и мир","Лев Толстой",400)

# Вывод информация о книге
print(my_book) # книга "Война и мир" автора Лев Толстой, 1225 стр.

# Проверим, является ли книга обьемной
print("Это обьемная книга?", my_book.is_long()) #Это обьемная книга ? True
```

```python
my_book.is_long()
```

```python
# Базовый класс
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        
    def start_engine(self):
        print(f"{self.brand} {self.model} двигатель заведен!") 
```

```python
# Производный класс
class Car(Vehicle):
    def __init__(self, brand, model, horsepower):
        super().__init__(brand, model) # Вызываем конструктор базового класса
        self.horsepower = horsepower
        
    def start_engine_car(self):
        print(f"{self.brand} {self.model} с помощью в  {self.horsepower} л.с. двигатель заведен!")
```

```python
vehicle = Vehicle("Toyota", "Corolla")
vehicle.start_engine() #Toyota Corolla двигатель заведен!
```

```python
# Создаем экземпляр производного класса
car = Car("Dodge", "Charger", 717)
car.start_engine() # Dodge charger с мощностью в 717 л.с. двигатель заведен!
```

```python
car.start_engine_car()
```
