```python
from scipy.stats import binom
import matplotlib.pyplot as plt
```

```python
# n - количество монеток , р - вероятность получить орла,  size - количество повторений эксперимента
results = binom.rvs(n = 10, p =0.5, size = 15, random_state = 42)
print(results)
```

```python
results.mean()
```

```python
plt.hist(results)
plt.show()
```
