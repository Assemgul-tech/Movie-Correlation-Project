```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-5, 5, 100)

plt.plot(x, (5-x)/3)
plt.plot(x, (5+x)/2)

plt.text(1,1.6, '$x_1+3x_2 = 5$')
plt.text(-3,0.5, '$x_1-2x_2 = -5$')

plt.grid(True, ls=':')
```

```python
plt.plot(x,(5-x)/3)
plt.plot(x, -x/3)

plt.text(1,1.6,'$x_1+3x_2=5$')
plt.text(0.3,-1.4, '$x_1+3x_2=0$')

plt.grid(True,ls=':')
```

```python
#сгенерируем случайные цифры аренды
np.random.seed(105)
random_numbers = [round(x, -3) for x in np.random.normal(150000, 15000, 1000)]
low_numbers = [round(x, -3) for x in np.random.uniform(60000, 90000, 50)]
high_numbers =[round(x, -3) for x in np.random.uniform(200000, 300000, 50)]

rent_prices = random_numbers+low_numbers+high_numbers
```

```python
print(f'средняя цена аренды : {np.mean(rent_prices)}')
print(f'медианная цена аренды : {np.median(rent_prices)}')
```

```python
print(f'стандартное отклонение цен составило: {np.std(rent_prices)}')
```

```python
min_price = min(rent_prices)
max_price = max(rent_prices)

print(f'минимальная цена аренды: {min_price}')
print(f'максимальная цена аренды: {max_price}')
print(f'размах цен: {(max_price - min_price)}')
```

```python
print(f'Первый квартиль: {np.percentile(rent_prices, 25)}')
print(f'Медиана: {np.percentile(rent_prices, 50)}')
print(f'Третий квартиль: {np.percentile(rent_prices, 75)}')
```

```python
# Урок 19 Как меняются квартили при изменении данных
```

```python
# пример 1: равномерное распределение 
np.random.seed(105)
data = np.random.uniform(0,100,10000).tolist()

q1 =np.percentile(data,25)
median = np.median(data)
q3 = np.percentile(data,75)

ticks = np.arange(0,110,10).tolist()

plt.hist(data,rwidth=0.9, bins=ticks)

plt.axvline(q1, color='red', linestyle='--',label='квартили')
plt.axvline(q3, color='red', linestyle='--')
plt.axvline(median, color='black', linestyle='--',label='median')
plt.legend()

plt.xticks(ticks)
plt.show()
```

```python
# пример 2: много маленьких значений
np.random.seed(105)
data = np.random.uniform(0,10,2500).tolist() + np.random.normal(50,13,7500).tolist()

q1 =np.percentile(data,25)
median = np.median(data)
q3 = np.percentile(data,75)

ticks = np.arange(0,110,10).tolist()

plt.hist(data,rwidth=0.9, bins=ticks)

plt.axvline(q1, color='red', linestyle='--',label='квартили')
plt.axvline(q3, color='red', linestyle='--')
plt.axvline(median, color='black', linestyle='-',label='median')

plt.legend()
plt.xticks(ticks)
plt.show()
```

```python
# пример 3: много больших значений
np.random.seed(105)
data = np.random.uniform(90,100,2500).tolist() + np.random.normal(50,15,7500).tolist()

q1 =np.percentile(data,25)
median = np.median(data)
q3 = np.percentile(data,75)

ticks = np.arange(0,110,10).tolist()

plt.hist(data,rwidth=0.9, bins=ticks)

plt.axvline(q1, color='red', linestyle='--',label='квартили')
plt.axvline(q3, color='red', linestyle='--')
plt.axvline(median, color='black', linestyle='-',label='median')

plt.legend()
plt.xticks(ticks)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Рассчитываем квантили
median = np.median(rent_prices)
q1 = np.percentile(rent_prices,25)
q3 = np.percentile(rent_prices,75)

# Рассчитываем IQR
iqr =q3 - q1

# Визуализация boxplot в сертикальном положении
plt.figure(figsize=(5,6))
plt.boxplot(rent_prices, vert=True, widths=0.7, patch_artist=False, medianprops={'color':'red'})

# Добавляем текстовыее метки для квартилей и IQR
plt.text(1.3,q1,'Q1',fontsize=10, color='black',fontweight='bold')
plt.text(1.3,median,'Median',fontsize=10, color='red',fontweight='bold')
plt.text(1.3,q3,'Q3',fontsize=10, color='black',fontweight='bold')
plt.text(1.1,q1-1.5*iqr,'Q1 -1.5*IQR',fontsize=10, color='blue',fontweight='bold')
plt.text(1.1, q3 + 1.5*iqr,'Q3 + 1.5*IQR',fontsize=10, color='blue',fontweight='bold')
         
# Добавляем заголовок и метки осей
plt.title('Bpxplot цен на аренду', fontsize=14)
plt.xticks([1],['Данные'])
plt.yticks(np.arange(50000, 320000, 20000))
plt.ylabel('Значения', fontsize=12)
         
# Отоброжаем график
plt.show()
```
